### Introduction: Linear model and an illustrative example

#### Illustrative example
Assume, we want to test a drug's efficacy by comparing it to a baseline. Let X be a new drug in clinical trials we want to test against a placebo to see if it lowers systolic blood pressure (SBP).

* $y$: The outcome, i.e. systolic blood pressure (measured in mmHg).
* $x$: The predictor, i.e. treatment assignment. We use the following coding:
    * $x = 0$ for patients receiving the placebo.
    * $x = 1$ for patients receiving drug X.

#### The Linear Model Equation
We represent this relationship using the standard linear equation:
$$y = \beta_0 + \beta_1x + \varepsilon$$

* $\beta_0$: The intercept represents the predicted SBP when $x=0$. Mathematically, this is the mean SBP of the placebo group.
* $\beta_1$ : The slope represents the effect of the treatment, i.e. the change in SBP when moving from placebo to Drug X. In a T-test, this is simply the difference between the two means.
* $\varepsilon$ : The error or "noise", e.g. measurement errors when taking BP or individual variation (not every patient reacts to the drug exactly the same way).

This representation transforms the "group comparison" into a "geometric line." Testing if the drug works is now mathematically identical to testing if the line connecting the two groups has a significant slope (i.e. is sloped as opposed to a horizontal line which has slope 0).

--- 
### 1. Hypothesis Mapping & Decision Rules: T-test vs. Linear Model

Let $\mu_{Drug}$ and  $\mu_{Placebo}$ be the mean SBP in the respective group. In a t-test, we focus on group means ($\mu$). In a linear model, we focus on the slope ($\beta_1$). They are mathematically identical ways of asking the same question.

#### The Null Hypothesis ($H_0$)
* T-test ($H_0$): $\mu_{Placebo} = \mu_{Drug}$
    * *Heuristic meaning:* "There is no difference in blood pressure between the sugar pill and the medicine."
* Linear Model ($H_0$): $\beta_1 = 0$
    * *Heuristic meaning:* If the slope is zero, the line is perfectly horizontal. This means the predicted blood pressure doesn't change regardless of whether you take the drug or the placebo.

#### The Alternative Hypothesis ($H_1$)
* T-test ($H_1$): $\mu_{Placebo} \neq \mu_{Drug}$
    * *Heuristic meaning:* "The drug causes a change in blood pressure."
* Linear Model ($H_1$): $\beta_1 \neq 0$
    * *Heuristic meaning:* The line is tilted. Whether it tilts down (lowers BP) or up (raises BP), a non-zero slope indicates that the treatment group ($x$) successfully predicts the outcome ($y$).

Both frameworks use the same "signal-to-noise" ratio ($t = \text{Estimate} / SE$) to decide whether to stick with the Null or move to the Alternative.

#### Accept vs. Reject $H_0$

To move from a guess to a decision, the linear model calculates a t-statistic. This is a ratio that compares your "signal" (the slope) to your "noise" (the standard error).

$$\text{t-statistic} = \frac{\text{Estimate } (\beta_1)}{\text{Standard Error } (SE)}$$
We follow the procedure:

1. Find the Slope ($\beta_1$): e.g., $-20$ mmHg.
2. Find the Standard Error ($SE$): e.g., $5$.
3. Calculate the t-statistic: $\text{Slope} / SE = -20 / 5 = -4$.
4. Convert to P-value: It looks at a distribution curve to see how much of the area sits beyond $-4$. A $t$ of $-4$ (depending on sample size) results in a p-value $ < 0.003$ (and represents the probability of seeing a the sloped line ($\beta_1=-20$) if the true effect were actually flat $\beta_1= 0$ , $\rightarrow$ i.e. drug doesn't help).
#### Rejecting $H_0$ (The drug works)
Reject the null if the "tilt" of your line is much larger than the "noise" of your data could cause. in that case the tilt is considered (statistically) significant (typically $|t| > 2$).
* The P-value Logic: The $p$-value represents the probability of seeing your tilted line if the true effect were actually flat  (e.g. for $p < 0.05$, that probability is too low to ignore).
* Linear Model Action: We reject the flat line. We conclude the slope ($\beta_1$) is statistically significant.
* Pharma Conclusion: "The drug effect is real; the signal drowned out the noise."

#### Accepting $H_0$ (The drug fails)
You accept the null (fail to reject) when the line is so close to horizontal that it could easily be a result of measurement error or biological variation ($\varepsilon$).
If the $t$-statistic is small (between $-2$ and $2$), meaning the slope is less than twice the size of the noise then (following our previous setting) $p \geq 0.05$, there is a greater than 5% chance that a perfectly flat line could have produced your results just by luck.
* Linear Model Action: We accept the flat line as a plausible reality.
* Pharma Conclusion: "The drug's effect is indistinguishable from placebo; the slope is effectively zero."

### 2. Mann-Whitney U: The Linear Model on Ranks

In pharmaceutical trials, the standard t-test can be compromised by extreme outliers (e.g., a patient with an anomalous reaction). The Mann-Whitney U addresses this by looking at the relative order $R = r_1,...,r_n$ of the data rather than raw values.

#### Rank Transformation
We transform every SBP value $y$ into its ranks $r_1,...,r_n$, where the lowest blood pressure in the entire study gets $r_1$, the second lowest $r_2$, and so on.
* The Model: $y_{rank}(x) = \beta_0 + \beta_1x + \varepsilon$
* $\beta_0$: The intercept represents the mean rank of the placebo group.
* $\beta_1$: The slope represents the difference in mean ranks between the placebo and Drug X.

#### Hypothesis Mapping
* Linear Model ($H_0$): $\beta_1 = 0$ (The mean rank is the same for both groups; the line is perfectly horizontal).
* Linear Model ($H_1$): $\beta_1 \neq 0$ (One group has a significantly higher or lower rank position, creating a tilted line).

#### The Decision Calculation
The procedure for "Accept vs. Reject" is identical to the t-test, only performed on the rank-transformed data:
1. Find the Rank Slope ($\beta_1$): e.g., $-15$ (The drug group dropped 15 rank positions on average).
2. Find the Standard Error of Ranks ($SE$): e.g., $6$.
3. Calculate the t-statistic: $-15 / 6 = -2.5$.
4. Conclusion: Since $|t| > 2$, we Reject $H_0$. The drug shifts the patient's rank position significantly, even if outliers make raw means unreliable.

### 3. Wilcoxon Signed-Rank: The Intercept-Only Model
This is used for "before and after" studies where each patient acts as their own control, effectively removing "biological noise" (baseline differences between patients). This is useful if there is massive biological variability between patients.

#### The "Drug vs. Placebo" (Crossover Design)
To fix the risk above, pharma uses a Crossover Trial. This is where "Drug" and "Placebo" become the "Before" and "After."

* The Setup:
    * Phase 1: Patient A takes the Placebo for a month. We measure BP.
    * Washout: Patient A takes nothing for two weeks to clear their system.
    * Phase 2: Patient A takes the Drug for a month. We measure BP.
* The Difference ($D$): $y_{Drug} - y_{Placebo}= y_{After} - y_{Before}$

The Calculation of $y_{sr}$:
1.  Compute Change ($D$): For each patient, $D = y_{After} - y_{Before}$.
2.  Rank Magnitude: Rank the absolute differences $|D|$.
3.  Assign Direction: Multiply the rank by the sign of $D$. This is $y_{sr}$.
    * *Example:* If Patient A (190 to 180) and Patient B (120 to 110) both drop the most, they both receive the same high negative rank, regardless of their different starting pressures.
We define $y_{sr}$ as the signed rank of the difference $D = y_{Drug} - y_{Placebo}$.

* The Model: $y_{sr} = \beta_0 + \varepsilon$
* $\beta_0$: The intercept representing the average signed rank of the population.

#### Hypothesis Mapping
* Linear Model ($H_0$): $\beta_0 = 0$ (The average internal shift is zero).
* Linear Model ($H_1$): $\beta_0 \neq 0$ (There is a statistically significant directed shift within patients).

#### The Decision Calculation
1. Estimate the Intercept ($\beta_0$): e.g., $12.5$.
2. Calculate the t-statistic: $t = \beta_0 / SE$. 
3. Conclusion: If $p < 0.05$, we Reject $H_0$, concluding a consistent treatment effect within the sampled patients.

### 4. The Wald Test: The Generalized Parameter Judge

Often times causation is multifactorial. To account for that with comparisons of complex models that account for multiple factors simultaneously (e.g., $y(x) = \beta_0 + \beta_1 \text{Drug} + \beta_2 \text{Age} + \beta_3 \text{Weight} + \varepsilon$) we can use the Wald test to judge if any specific factor ($\beta_i$) is actually contributing to the model.

The Wald statistic ($W$) is essentially the squared version of the signal-to-noise ratio. It measures how many standard deviations the estimated $\beta_i$ is away from zero.
$$W = \frac{\hat{\beta}_i^2}{Var(\hat{\beta}_i)} = t^2$$

#### Hypothesis Mapping
* Null Hypothesis ($H_0$): $\beta_i = 0$
    * *Heuristic meaning:* "This specific factor (e.g., Age) has zero effect on blood pressure in our model."
* Alternative Hypothesis ($H_1$): $\beta_i \neq 0$
    * *Heuristic meaning:* "This factor is a significant predictor and must remain in the model."

#### The Decision Calculation
The procedure follows the same "signal vs. noise" logic, but uses a Chi-square ($\chi^2$) distribution for the final decision:

1.  Estimate the Parameter: e.g., $\beta_{Age} = 0.5$ (meaning for every year of age, SBP increases by 0.5 mmHg).
2.  Calculate the Variance ($Var$): The uncertainty associated with that age estimate.
3.  Compute $W$: Square the estimate and divide by the variance. 
4.  Accept vs. Reject:
    * If $W > 3.84$ (which corresponds to the standard $|t| > 1.96$), then $p < 0.05$.
    * Action: We Reject $H_0$.
    * Conclusion: "Age is a statistically significant predictor of blood pressure; the effect we see is too large to be attributed to noise."

### Comparison

| Traditional Test | Linear Model Equation | Meaning of the signal ($\beta$) | Decision Logic |
| :--- | :--- | :--- | :--- |
| Student's t | $y(x) = \beta_0 + \beta_1x + \varepsilon $ | Difference in raw means | Reject $H_0$ if $\beta_1 \neq 0$ |
| Mann-Whitney | $y_{rank}(x) = \beta_0 + \beta_1x + \varepsilon$ | Difference in mean ranks | Reject $H_0$ if $\beta_1 \neq 0$ |
| Wilcoxon | $y_{sr} = \beta_0 + \varepsilon$ | Mean signed rank | Reject $H_0$ if $\beta_0 \neq 0$ |
| Wald Test | $\beta_i / SE$ (the $\varepsilon$ is in the denominator here) | Significance of the i-th factor | Reject $H_0$ if $\beta_i \neq 0$ |
